Langkah 1 - Memahami Koleksi Dokumen

```python
koleksi_dokumen = {
    1: "Sistem temu kembali informasi digunakan untuk mencari data di internet",
    2: "Jaringan komputer MANET menggunakn protokol routing OLSR",
    3: "Mikrotik adalah router yang banyak digunakan di kampus",
    4: "Artificial intelligence dan Agentc AI menjadi tren di tahun 2026",
    5: "Database berbeda dengan sistem temu kembali karena hasilnya pasti"
}

print("Jumlah dokumen dalam koleksi: ", len(koleksi_dokumen))
```

Langkah 2 - Preprocessing: Membersihkan Teks

```python
def preprocessing(teks):
  teks = teks.lower()     #1. ubah seluruhnya menjadi huruf
  tokens = teks.split()   #2. pecah menjadi kata per kata
  return tokens

#print(preprocessing("Protokol Routing OLSR"))
print(preprocessing("Mikrotik Router Kampus"))
```

Langkah 3 - Membuat Fungsi Pencarian Sederhana

```python
def cari_dokumen(query, koleksi):
  query_tokens = preprocessing(query)   #1. bersihkan kueri
  hasil = []

  for doc_id, doc in koleksi.items():   #2. periksa tiap dokumen
    doc_tokens = preprocessing(doc)
    skor = len(set(query_tokens) & set(doc_tokens))   #3. kata sama

    if skor > 0:    #4. simpan bila cocok
      hasil.append((doc_id, skor, doc))

  hasil.sort(key = lambda x:[1], reverse=True) #5. urutkan skor
  return hasil
```

Langkah 4 - Uji Coba dengan Tiga Kueri

```python
def tampilan(query, koleksi):
  print(f"Query: '{query}')")
  hasil = cari_dokumen(query, koleksi)
  if not hasil:
    print(" (tidak ada dokumen yang cocok)")
  for doc_id, skor, doc in hasil:
    print(f"  Dokumen {doc_id} | Skor: {skor} | {doc}")
  print()

tampilan("protokol routing", koleksi_dokumen)     #Uji 1
tampilan("database pasti", koleksi_dokumen)
tampilan("sistem temu kembali", koleksi_dokumen)  #Uji 2
tampilan("AI 2026", koleksi_dokumen)
tampilan("router kampus", koleksi_dokumen)        #Uji 3
```
